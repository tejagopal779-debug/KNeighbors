```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\DELL\Downloads\temperture\diabetes_prediction_dataset.csv")
```

```python
df
```

```python
x=df.drop(columns="diabetes")
y=df["diabetes"]
```

```python
x_num=x.select_dtypes(include="number")
x_cat=x.select_dtypes(exclude="number")
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
st=StandardScaler()
x_num=st.fit_transform(x_num)
```

```python
on=OneHotEncoder()
x_cat=on.fit_transform(x_cat).toarray()
x_final=np.hstack((x_num,x_cat))
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
x_ta, x_te, yU, Y_u=train_test_split(x_final,y,test_size=0.12)
```

```python
knn=KNeighborsClassifier(n_neighbors=3)
u=knn.fit(x_ta,yU)
```

```python
per=knn.predict(x_te)
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(Y_u,per)
```
